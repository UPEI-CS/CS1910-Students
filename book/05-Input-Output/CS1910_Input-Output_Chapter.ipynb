### Console Input and Output

The *console* is the default location for text input and output from/to the computer's user.  Console output goes to the screen (usually a terminal window).  Console input comes from the keyboard.

#### Outputting Text to the Screen

We have seen that when we use Python in interactive mode and enter an expression, Python responds with the value of the expression.  But when we use Python in non-interactive mode, this is not the case.  Suppose we put the following Python program into a file called `arithmetic.py`:

`arithmetic.py`
```python
pi = 3.14159
r = 7
pi * r**2
2 / 7 - 12
2 / (7 - 12)
```

Then suppose we run this program:

```text
$ python arithmetic.py
$
```

Nothing happened!  Or at least it **appears** that nothing happened because Python didn't output any responses.  All of the computations in the program **did** occur, but nothing was printed out in response.  In a non-interactive program we have to explicitly ask Python to print values to the console.  We do this using the `print()` syntax.  Here we have modified the program to print out the values of the three expressions:

````python
pi = 3.14159
r = 7
print( pi * r**2 )
print( 2 / 7 - 12 )
print( 2 / (7 - 12) )
````

To use the `print()` syntax we type the word `print`, followed by whatever expression whose value we want printed enclosed in a pair of parentheses.  When we run the modified program, we now get some results:

```text
$ python arithmetic.py
153.93791
-11.714285714285714
-0.4
```

You can print the values of more than one expression at once by providing a comma-separated list of expressions within the parentheses. Each value printed in this manner is separated by a space character.  This allows you to combine data from several literals or variables to produce a single message.  Here's a program stored in `printStuff.py`:

````python
print("Two to the power of six is: ", 2 ** 6)
a = 8
b = 5
print("The remainder after dividing", a, "by", b, "is:", 8 % 5)
````

And here is its output:

```

iroh:CS1910 chris$ python printStuff.py 
Two to the power of six is:  64
The remainder after dividing 8 by 5 is: 3

```

> **Know Python 2?**
>
> The `print` syntax is different in Python 3 compared to Python 2.  In Python 2 the parentheses are optional because `print` was a statement, not a function.  In Python 3, they are needed because `print` is a function in Python 3.

#### Reading Strings from the Keyboard

You can ask for input from the keyboard with the `input()` syntax:

````python
# ▶ Interactive: this cell waits for you to type something.
# Run it yourself - "Run All" skips it so the rest of the chapter still works.
x = input()   
````

This line will pause the program, and wait for the user to type something and press the Enter key.  Whatever the user typed will be stored as a string associated with the variable `x`.  You can optionally ask for Python to print out a prompt to the user by providing a string inside of the parentheses.  Here's an example:

````python
# ▶ Interactive: this cell waits for you to type something.
# Run it yourself - "Run All" skips it so the rest of the chapter still works.
x = input("Please enter your name: ")  
print("Hello,", x) 
````

Here's what happens when we run this:

```text
$ python hello.py
Please enter your name: Mark
Hello, Mark
```

*(`Mark` is what the user typed; everything else was printed by the program.)*

The line `x = input('Please enter your name: ')` first prints the string provided, then it waits for the user to type text and press enter.  The bright red text was typed by the user, and this text is given the variable name `x`.  The program responds by printing out a greeting using the text that was entered.

#### Reading Numbers from the Keyboard

Reading numbers is a bit trickier because `input()` will only read strings.  However, we can convert the string to a number, if the string is actually a representation of a number.

````python
# ▶ Interactive: this cell waits for you to type something.
# Run it yourself - "Run All" skips it so the rest of the chapter still works.
# read an integer
x = int(input("Enter an integer: "))                 

# read a floating-point number
y = float(input("Enter a floating-point number: "))  
````

The text that the user typed for the first `input` will be converted to an integer and given the name `x`, if that text contains only digits.  Otherwise, Python will issue an error.  The second `input` will convert the typed text to a floating-point number and give it the name `y`, if the text is a valid representation of a floating point number.  Again, if it isn't, Python will issue an error.

### More About Commenting

You have seen in this chapter some initial very simple programs.  You will note though that we have not provided any comments to these primarily to keep them readable in the text.

However, as we mentioned, such notes can be very valuable in understanding what code is doing.  Consider the following program:

````python
pi = 3.14159
r = 7
print( pi * r**2 )
print( 2 / 7 - 12 )
print( 2 / (7 - 12) )
````

It would be very tempting, and many early programmers would add comments like so:

````python
# Set a variable pi to 3.14159
pi = 3.14159
# Set a variable r to 7
r = 7
#Print the value of pi times r squared
print( pi * r**2 )
#Print the value of 7 minus 12 and divide 2 by the result.
print( 2 / 7 - 12 )
````

However, this type of commenting actually contravenes the purpose of commenting in the first place.  The purpose of commenting is to help other programmers (or yourself) understand *why* the program was written and what it is supposed to do.  The above comments document *what* each specific line does.  Essentially, these comments are simply translating the formal language of programming into natural language, which in the long run is not very helpful.

While there is no specific rules on how to do this well, you should be looking to document what code is doing in the *context of the program*.  You need to pull your comments up to a level of abstraction so that other programmers can understand how a section of code works in reference to the program as a whole.

For example, with the above code, there are two distinct parts of the program.  The first part sets some variables for calculating attributes of circles, while the second prints material.  We can adjust the comments in the following way:

````python
# Create variables for performing calculations about a circle
pi = 3.14159 #the constant pi
r = 7 #the radius of the circle

#Print a number of calculations for display to the user
print( pi * r**2 )
print( 2 / 7 - 12 )
````

Of course this is a very simplistic program.  Longer programs may need more information for other blocks of code.  As a final note, you will find it useful to document with comments what specific variables contain for future reference when you are working on your code.

### More Built-In Python Functions

We finish with a few of the most commonly used built-in Python functions.  Run
each example, and try changing the arguments to see what happens.

#### `max`

Takes **two or more** arguments and returns the largest of them.

````python
largest = max(10, 15, 42, 19)
print(largest)
````

#### `min`

Works like `max`, but returns the *smallest* of its arguments.  It also accepts
any number of arguments.

#### `abs`

Takes **one** argument and returns its absolute value.

````python
print(abs(-18))
````

#### `round`

Takes **one required** argument and **one optional** one.  It returns the number
rounded to the given number of decimal places; with no second argument it rounds
to the nearest whole number.

````python
print(round(18.234))
print(round(18.274, 1))
````

#### `pow`

Takes **two** arguments.  `pow(x, y)` returns `x` raised to the power of `y`.

````python
print(pow(7,2))
print(pow(5.0, 1/2))
````

#### `type`

Takes **one** argument and returns its data type.

````python
type (5)
type (2.3)
type("foo")
````

#### `int`

Takes **one** argument and converts it to the integer data type, if that is
possible.  A string can be converted if it contains only the digits 0-9 and
possibly a decimal point.

````python
int("42")
````

#### `float`

Like `int`, but converts its argument to the floating-point data type.

#### `str`

Like `int`, but converts its argument to the string data type.

the function then waits for the user to type something and press enter, and
returns the text that was entered.  This is the same function used for console
input in the Input/Output chapter.
