## What this lab practises

*Containers* are objects that store other objects: a `list` stores a sequence of objects. Containers can also hold other containers, which lets you represent data of any shape, such as a grid of rows and columns. In this lab you will store a game board as a list of lists, and process it with nested loops.

### Key points to take away

- A **list of lists** can store a grid. `grid[r]` is one row (a list); `grid[r][c]` is one cell. The first index is the row, the second the column.
- `len(grid)` is the number of rows; `len(grid[0])` is the number of columns.
- `pprint.pprint(...)` ("pretty print") prints a list of lists one row per line, which is much easier to read.
- A **nested loop** has a loop inside a loop. The inner body runs (outer repetitions) × (inner repetitions) times.
- You can loop over the items (`for row in grid:`) or over the indexes (`for i in range(len(grid)):`). Use indexes when you need the position, or need to change the list.

````python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math as _math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return _math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
````

## Part 2: Nested `for` loops (about 20 minutes)

We usually use a `for` loop to look at the contents of a list:

```python
groceries = ["Apples", "Milk", "Flour", "Chips"]
for item in groceries:
    print("Don't forget the", item)
```

If a list contains other lists, we need a **nested** `for` loop. This code counts the empty `' '` cells in `grid`:

```python
count = 0
for row in grid:          # outer loop
    print("row =", row)
    for cell in row:      # inner loop
        print("cell =", cell)
        if cell == ' ':
            count += 1
print(count, "spaces remaining")
```

(`count += 1` is a short way of writing `count = count + 1`.)
