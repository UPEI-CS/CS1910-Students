### 3.3 Run the tests

Here, at last, is a programmer's attempt at `password_ok`. It is followed by a small **test driver**: a function `test` that runs `password_ok` on one input and reports `PASS` or `FAIL`.

Run the cell. Then add a `test(...)` line for each of your own test cases from 3.2, question 5, and run it again.

````python
def password_ok(pw):
    has_digit = False
    for i in range(len(pw) - 1):
        if pw[i].isdigit():
            has_digit = True
    return len(pw) > 8 and has_digit


def test(pw, expected):
    actual = password_ok(pw)
    if actual == expected:
        print("PASS", f'"{pw}"')
    else:
        print("FAIL", f'"{pw}"', "expected", expected, "but got", actual)


test("sunshine42", True)
test("sunshine", False)
test("abc123", False)
test("", False)
test("12345678", True)
test("password9", True)
test("abcdefg1", True)

# Add your own tests from 3.2 here:
````

### 4.4 Using the debugger

Adding `print` calls works, but you must decide in advance what to print, and remember to remove them afterwards. A **debugger** lets you *pause* a running program and look at any variable you like.

The cell below calls `breakpoint()` inside the loop. When Python reaches that line it pauses, and a box appears where you can type debugger commands:

| Command | What it does |
|---------|--------------|
| `p ch` | **p**rint the value of the variable `ch` (use any variable name) |
| `n` | run the **n**ext line, then pause again |
| `c` | **c**ontinue running until the next `breakpoint()` |
| `q` | **q**uit the debugger and stop the program |

Run the cell. At each pause, type `p ch` and `p total`, then `c` to go around the loop again. When you have seen enough, type `q`.

⚠️ **Always finish with `q`** (or keep pressing `c` until the program ends). While the debugger is waiting for you, no other cell can run.

````python
def sum_digits(text):
    for ch in text:
        total = 0
        breakpoint()          # the program pauses here every time around the loop
        if ch.isdigit():
            total = total + int(ch)
    return total


print("Result:", sum_digits("507"))
````
