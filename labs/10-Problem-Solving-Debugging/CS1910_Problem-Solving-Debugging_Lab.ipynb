## What this lab practises

Every programmer writes bugs. What makes a good programmer is having a *process* for finding and fixing them. In this lab you will sort errors into kinds, read the messages Python gives you, design tests that catch bugs, and use `print` and the debugger to see inside a running program. Then you will use the whole four-stage process on a new problem.

### Key points to take away

- **Syntax errors** break Python's grammar. They are found *before* the program runs, so nothing runs at all.
- **Runtime errors** are found *while* the program runs, at the line that cannot be done. The lines before it have already run.
- **Semantic (logic) errors** give no message: the program runs, but the answer is wrong. Only **testing** finds them.
- Read a **stack trace** from the bottom: the last line names the error, the bottom block is where it happened, and the blocks above show the chain of calls. The line that **crashes** is not always the line that is **wrong**.
- A **test case** is an input plus the **expected output**, worked out from the specification *before* you run the code. Include **normal**, **edge** and **boundary** cases.
- **Debugging by printing:** add `print` calls to watch variables change. **Debugger:** `breakpoint()` pauses the program; `p` prints a variable, `n` runs the next line, `c` continues, `q` quits.
- When testing shows a gap in your design, going back to the Design stage is a normal part of the process, not a failure.

```python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
```

### B3. Debug by printing

`sum_digits` should return the sum of all the digits in a string, ignoring other characters. For example, `sum_digits("a1b2c3")` should be `6`. It gives wrong answers, and the last check even crashes.

The cell already has a **debugging print** inside the loop that shows `ch` and `total` each time around. Run it, read the output line by line, and answer in the ✏️ cell:

4. What **single** change fixes both the wrong answers *and* the error?

Then fix the function, and delete the debugging print.

```python
def sum_digits(text):
    for ch in text:
        total = 0
        if ch.isdigit():
            total = total + int(ch)
        print(f"  ch = '{ch}'   total = {total}")    # debugging print: delete it when you are done
    return total

print("Result:", sum_digits("123"))
print("Result:", sum_digits("a1b2c3"))

# ---- Checks: run the cell to test your code (do not change below) ----
check(sum_digits("123"), 6, 'sum_digits("123")')
check(sum_digits("a1b2c3"), 6, 'sum_digits("a1b2c3")')
check(sum_digits("507"), 12, 'sum_digits("507")')
check(sum_digits("x9y"), 9, 'sum_digits("x9y")')
check(sum_digits(""), 0, 'sum_digits("")')
```

### B4. Using the debugger

Adding `print` calls works, but you must decide in advance what to print, and remember to delete them afterwards. A **debugger** lets you *pause* a running program and look at any variable you like.

The cell below has the **original, buggy** `sum_digits` (renamed `buggy_sum_digits`), with a `breakpoint()` inside the loop. When Python reaches that line it pauses, and a box appears where you type debugger commands:

| Command | What it does |
|---|---|
| `p ch` | **p**rint the value of the variable `ch` (use any variable name) |
| `n` | run the **n**ext line, then pause again |
| `c` | **c**ontinue running until the next `breakpoint()` |
| `q` | **q**uit the debugger and stop the program |

Run the cell. At each pause, type `p ch` and `p total`, then `c` to go around the loop again. When you have seen enough, type `q`.

⚠️ **Always finish with `q`** (or keep pressing `c` until the program ends). While the debugger is waiting for you, no other cell can run.

```python
def buggy_sum_digits(text):
    for ch in text:
        total = 0
        breakpoint()          # the program pauses here every time around the loop
        if ch.isdigit():
            total = total + int(ch)
    return total

print("Result:", buggy_sum_digits("507"))
```
