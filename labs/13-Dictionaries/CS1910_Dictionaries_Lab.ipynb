```python
#@title ▶️ Run this cell first
#@markdown This practice checker reports which tests pass. You don't need to edit it.

def check(got, expected, label):
    if type(got) is type(expected) and got == expected:
        print(f"✅ {label}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")

print("✅ Practice checker ready.")
```

```python
# 1.3 Predict first, then run.
# My prediction:

try:
    print(elements[N])
except (NameError, KeyError) as error:
    print(type(error).__name__, error)

try:
    print(elements['nitrogen'])
except (NameError, KeyError) as error:
    print(type(error).__name__, error)

try:
    print(elements[1])
except (NameError, KeyError) as error:
    print(type(error).__name__, error)
```
