## What this workbook reviews

A **loop** repeats a block of code. A `for` loop repeats once for each item in a sequence, such as the numbers in a `range`. A `while` loop repeats *as long as* a condition stays `True`. This workbook practises predicting how many times a loop runs, tracing variables through a loop, the counter and accumulator patterns, sentinel loops, nested loops, and avoiding infinite loops.

### Key points to take away

- `range(start, stop, step)` **stops before** `stop`. `range(5)` is 0, 1, 2, 3, 4, which is five values. `range(10, 0, -3)` counts down: 10, 7, 4, 1.
- A `while` loop needs three parts: **initialize** a variable before the loop, **test** it in the condition, and **update** it in the body. Forget the update and you get an **infinite loop**. (Press the ■ stop button to interrupt one.)
- A `while` loop's body may run **zero times** if the condition is `False` from the start.
- **Counter pattern:** `count = 0` before the loop, `count += 1` inside an `if` in the loop.
- **Accumulator pattern:** `total = 0` before the loop, `total += value` inside the loop.
- A **sentinel loop** keeps going until a special value appears (for example, a valid input). After the loop, you *know* the condition is `False`.
- **Nested loops:** the inner loop runs completely for *each* repetition of the outer loop, so the total is outer × inner.
- Use `for` when you know how many times to repeat (or what to loop over), and `while` when you repeat *until* something happens.

```python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
```
