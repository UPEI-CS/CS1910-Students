## What this workbook reviews

A **variable** is a name that refers to a value. A **literal** is a value written directly in your code, such as `42`, `3.5` or `"hello"`. Every value has a **type** that decides what you can do with it. This workbook practises assignment and reassignment, the five basic types, converting between types, naming rules, and a first look at repeating code with `for` and `range`.

### Key points to take away

- `x = 5` is read "x **gets** 5". The right-hand side is worked out **first**, then the name on the left is attached to the result. So `x = x + 2` is perfectly sensible.
- A variable must be assigned **before** it is used, or you get a `NameError`.
- The basic types: `int` (`7`), `float` (`7.0`), `str` (`"7"`), `bool` (`True`/`False`) and `None`. `type(value)` tells you which.
- `"7" + "3"` is `"73"` (joining text), while `7 + 3` is `10`. The type decides what `+` means.
- Convert with `int()`, `float()`, `str()` and `bool()`. `int("3.5")` is a `ValueError`, but `int(3.9)` is `3` (it chops off the decimal part; it doesn't round).
- Names: use `snake_case`, make them meaningful, never start with a digit, and never use a reserved word (`class`, `for`, `if`, …).
- `b = a` makes `b` refer to the same value. Reassigning `a` afterwards does **not** change `b`.
- `for i in range(n):` repeats the indented block `n` times, with `i` taking the values 0, 1, …, n − 1.

```python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
```
