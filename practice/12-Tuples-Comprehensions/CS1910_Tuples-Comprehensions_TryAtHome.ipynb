## What this workbook reviews

A **tuple** is like a list that can't be changed: an immutable sequence, useful for fixed groups of values such as a `(row, col)` position or a function returning several results. A **2D list** is a list of lists, which gives you a grid of rows and columns. A **list comprehension** builds a list in a single readable line. This workbook practises all three.

### Key points to take away

- Tuples: `point = (3, 4)`. You index them like a list, but `point[0] = 9` is a `TypeError`. A one-item tuple needs a comma: `(5,)`. Plain `(5)` is just the number 5.
- **Unpacking:** `x, y = point`. A function can `return a, b`, which returns the tuple `(a, b)`, and the caller can unpack it.
- Use a tuple when the group of values is fixed and shouldn't change. Use a list when items will be added, removed or changed.
- A 2D list `grid[r][c]`: the **row** comes first, then the column. `len(grid)` is the number of rows, and `len(grid[0])` is the number of columns.
- Traverse a grid with nested loops. Use **for-each** (`for row in grid:`) when you only need values, and **index-based** (`for r in range(len(grid)):`) when you need positions or neighbours.
- A comprehension `[expression for item in iterable if condition]` is the same as an `append` loop, written in one line.
- Sorting a list of lists or tuples is **lexicographic**: compare the first items, and if they tie, compare the second items, and so on.

```python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
```
