## What this workbook reviews

A **list** is an ordered sequence of values, like a string. Unlike a string, a list is **mutable**: you can change, add and remove items in place. Mutability makes lists powerful, but it is also behind the most common list surprises: methods that return `None`, two names that refer to the same list, and functions that change the list you pass in. This workbook practises all of these.

### Key points to take away

- Index and slice lists just like strings: `items[0]`, `items[-1]`, `items[1:3]`. You can also **assign** to an index: `items[0] = 99`.
- Methods that change the list in place return **`None`**: `append`, `sort`, `reverse`, `clear`. Writing `nums = nums.sort()` loses your list! Use `nums.sort()` on its own line.
- `b = a` does **not** copy a list. Both names refer to the *same* list, so a change through one shows up through the other. Use `b = a.copy()` (or `a[:]`) for a separate copy.
- A list passed into a function is the *same* list, so if the function changes it, the caller's list changes too. Copy it first if the function shouldn't change its input.
- A **shallow copy** copies only the outer list. Nested lists inside it are still shared.
- `x in items` / `x not in items` test membership. `items.count(x)` and `items.index(x)` search.
- `len`, `sum`, `max` and `min` work on lists. `a + b` joins two lists into a new one, and `[0] * 3` is `[0, 0, 0]`.
- Build a list with a loop: start with `result = []` and `result.append(...)` inside the loop.

```python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
```
