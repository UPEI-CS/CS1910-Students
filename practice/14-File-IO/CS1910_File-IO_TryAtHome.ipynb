## What this workbook reviews

Variables live in memory and vanish when your program stops. **Files** keep data on disk. This workbook practises reading a file line by line, cleaning and converting what you read, splitting delimited (CSV) data, writing and appending to files, and joining values together for output. The setup cell creates the small data files that the exercises use, so this notebook works on its own.

### Key points to take away

- `f = open("data.txt")` opens a file for reading (`"r"` is the default). Always `f.close()` when you are done, or let `with open("data.txt") as f:` close the file for you automatically.
- `for line in f:` reads one line at a time. Every line keeps its trailing **`\n`**, so use `line.strip()` before comparing or converting it.
- Everything read from a text file is a **string**. Convert it with `int()` or `float()` before doing arithmetic.
- `f.readlines()` (or `list(f)`) reads all of the lines into a list at once.
- `line.split(",")` breaks one line of CSV data into a list of strings.
- `open(name, "w")` **creates or overwrites** a file, and `open(name, "a")` **appends** to the end. `f.write(text)` does **not** add a newline, so you add `\n` yourself.
- `", ".join(list_of_strings)` glues strings together with a separator. To join numbers, convert them first: `"\n".join([str(n) for n in nums])`.
- Paths: a **relative** path (`"data/marks.csv"`) starts from the current folder, and an **absolute** path starts from the root of the file system. `os.path.join("data", "marks.csv")` builds a path with the right separator for the operating system.

````python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")

# Create the small data files this workbook uses (safe to run again)
with open("colors.txt", "w") as f:
    f.write("red\ngreen\nblue\n")
with open("numbers.txt", "w") as f:
    f.write("10\n25\n7\n8\n")
with open("temps.txt", "w") as f:
    f.write("12.5\n\n-3.0\n8.25\n\n")
with open("marks.csv", "w") as f:
    f.write("name,mark\nAna,85\nBen,92\nCy,78\nDee,92\n")
with open("sales.csv", "w") as f:
    f.write("date,item,qty,price\n2026-10-01,muffin,3,2.50\n2026-10-01,coffee,2,1.75\n"
            "2026-10-02,muffin,1,2.50\n2026-10-02,tea,4,1.25\n")
print("✅ Data files created: colors.txt, numbers.txt, temps.txt, marks.csv, sales.csv")
````
