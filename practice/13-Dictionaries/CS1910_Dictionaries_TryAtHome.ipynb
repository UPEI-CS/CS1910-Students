## What this workbook reviews

A **dictionary** maps **keys** to **values**: you look values up by name rather than by position. Dictionaries are the natural tool for counting things, for translating one thing into another, and for storing records. This workbook practises creating and updating dictionaries, safe lookups, the iteration patterns, nested data, dictionary comprehensions, and conditional expressions.

### Key points to take away

- `d = {"Ana": 19, "Ben": 21}`. `d[key]` looks up a value, and a missing key is a `KeyError`. `d.get(key, default)` returns the default instead of raising an error.
- `d[key] = value` **adds** the key if it is new, or **replaces** the value if the key already exists. Keys are unique.
- `del d[key]` or `d.pop(key)` removes a key. `key in d` tests whether a key is present.
- Keys must be **immutable** (str, int, tuple); a list can't be a key. Values can be anything, including lists and other dictionaries.
- Iterating: `for k in d:` gives the keys; `for v in d.values():` the values; `for k, v in d.items():` both. `.keys()`, `.values()` and `.items()` return *views* that reflect later changes to the dictionary.
- **The counting pattern:** `counts[x] = counts.get(x, 0) + 1`.
- A dictionary comprehension: `{k: v for item in iterable if condition}`.
- A conditional expression: `"odd" if n % 2 else "even"`.

````python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
````
