## What this workbook reviews

A **function** packages a task under a name, so you can use it again without repeating code. When you call a function, you hand it **arguments**, which are assigned to its **parameters**. When it finishes, it **returns** a value to the code that called it. This workbook practises the difference between printing and returning, scope, default and keyword arguments, generalizing code into parameters, and building small functions that work together.

### Key points to take away

- `def name(parameters):` *defines* a function. Nothing in the body runs until the function is **called**, and it must be defined *before* it is called.
- `print` shows a value to a **person**. `return` gives a value back to the **program**, so it can be stored, tested or used in a calculation.
- A function with no `return` statement (or with a bare `return`) returns `None`.
- `return` ends the function **immediately**, even in the middle of a loop.
- Variables assigned inside a function are **local**: they exist only during that call, and they do not change variables with the same name outside it.
- **Default arguments** (`rate=0.15`) make a parameter optional. **Keyword arguments** (`price_with_tax(rate=0, price=20)`) let you name parameters when you call the function.
- **Generalize** by turning a fixed value into a parameter: `count_spaces(text)` becomes `count_char(text, ch)`.
- A good function does **one clear job** (cohesion). It has a descriptive name, a docstring, and type hints such as `def area(w: float, h: float) -> float:`.

```python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
```
