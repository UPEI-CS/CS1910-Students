## What this workbook reviews

Writing code is only one stage of solving a problem. This workbook practises the whole process: **Understand → Design → Implement → Evaluate**. It also practises the debugging skills you need when the Evaluate stage finds a problem: recognising the three kinds of error, reading a stack trace, designing tests that expose bugs, and using inspection and `print` to find them.

### Key points to take away

- **Understand:** what is the desired outcome, what information is given, and what is missing? Work one example by hand.
- **Design:** write pseudocode *before* Python. **Implement:** translate it, one small piece at a time. **Evaluate:** test it. When testing exposes a gap in the design, go back and iterate.
- Good tests include a **normal case**, an **edge case** (empty input, a single item, zero) and a **boundary case** (exactly on a threshold).
- **Syntax errors** stop the program before it runs (a missing quote, colon or bracket). **Runtime errors** happen while it runs, and come with a stack trace (`IndexError`, `ZeroDivisionError`, `ValueError`, …). **Semantic errors** run without complaint but give the *wrong answer*; only tests catch these.
- Read a **stack trace** from the **bottom up**: the last line names the error, and the lines above it show the chain of function calls that led there.
- **Debugging by inspection:** read the code and trace it by hand. **Debugging by printing:** add `print(...)` calls to show variable values at key points. You can also use `breakpoint()` to pause execution and look around.
- Indentation bugs are common semantic errors. A `return` indented one level too far, inside a loop, is the classic example.

````python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
````

### B3. Reading a stack trace

Running `report(3, 0)` gives this stack trace:

```
ZeroDivisionError                         Traceback (most recent call last)
----> 9 print(report(3, 0))
----> 6     return f"{percent(part, whole):.1f}%"
----> 2     return part / whole * 100
ZeroDivisionError: division by zero
```

Which function did the error happen in? Which call caused it? Fix `percent` so that it returns `0.0` when `whole` is 0.

````python
def percent(part, whole):
    return part / whole * 100


def report(part, whole):
    return f"{percent(part, whole):.1f}%"

# ---- Checks: run the cell to test your code (do not change below) ----
check(percent(1, 4), 25.0, "percent(1, 4)")
check(percent(3, 0), 0.0, "percent(3, 0)")
check(report(3, 0), "0.0%", "report(3, 0)")
````
