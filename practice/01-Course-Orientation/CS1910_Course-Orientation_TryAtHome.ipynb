## What this workbook reviews

Everything in this course happens in **notebooks** (Google Colab or Jupyter). A notebook mixes **text cells** (like this one) with **code cells** that run Python. This first workbook gets you comfortable running cells, reading what Python prints, and treating error messages as useful information rather than failures. It also introduces the habit this course values most: **plan the steps before you write the code.**

### Key points to take away

- Run a code cell with the ▶️ button or **Shift + Enter**. The output appears underneath it.
- Cells run in the order **you run them**, not the order they appear on the page. If something seems wrong, use *Runtime → Restart and run all* to run everything from the top.
- `print(...)` displays a value. Text inside quotes (`"Hello"`) is printed exactly as written, while `print(2 + 3)` works out the sum first and prints `5`.
- Anything after a `#` is a **comment**. Python ignores it, so it is a note for humans.
- **Error messages are information.** The last line tells you the *kind* of error (`SyntaxError`, `NameError`, …), and the arrow or line number tells you *where*. Read them, don't just rerun.
- Before coding, describe the steps in plain language. A clear plan makes the code much easier to write.

````python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
````
