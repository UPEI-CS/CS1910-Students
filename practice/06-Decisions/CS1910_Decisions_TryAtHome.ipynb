## What this workbook reviews

Programs normally run one line after another. A **decision** (`if` / `elif` / `else`) lets a program choose which lines to run, based on a condition that is either `True` or `False`. This workbook revisits how Python chooses a branch, why the order of conditions matters, how to test boundaries, and how to turn a real-world rule into clear conditional code.

### Key points to take away

- In an `if` / `elif` / `else` chain, Python checks the conditions **from top to bottom** and runs **only the first** branch whose condition is `True`. The rest are skipped.
- Separate `if` statements are **each** checked, so more than one of them can run.
- When conditions overlap, put the **most specific** (narrowest) condition first.
- Indentation decides what belongs to a branch. A line indented less than the branch always runs.
- Combine conditions with `and`, `or` and `not`. A chained comparison such as `13 <= age <= 64` reads like maths.
- **Test the boundaries.** If a rule says "50 or more", test 49.99, 50 and 50.01. Most decision bugs are `>` where `>=` was meant, or the other way around.
- Never compare floats with `==`. Use `math.isclose(a, b)` or `abs(a - b) < 1e-9`.
- Strings compare by character code, so `"Zebra" < "apple"` is `True` because every uppercase letter comes before every lowercase letter.
- Write `if is_member:` rather than `if is_member == True:`.

```python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
```
