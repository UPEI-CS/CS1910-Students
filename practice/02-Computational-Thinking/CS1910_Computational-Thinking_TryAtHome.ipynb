## What this workbook reviews

**Computational thinking** is a way of approaching problems so that a computer (or a person) can solve them step by step. Its parts are **decomposition** (break the problem into smaller parts), **pattern recognition** (notice repeated structure), **abstraction** (keep only the details that matter), **algorithms** (precise steps, often written as **pseudocode**), and **validation** (check that the result is correct). This workbook practises those skills with Turtle drawings and everyday problems.

### Key points to take away

- **Decomposition:** a house is a square plus a triangle. Solve the small parts, then combine them.
- **Pattern recognition:** a square is "forward, turn" four times. Spotting the repeat is what later lets you write a loop.
- **Abstraction:** ignore what doesn't matter (the colour of the cards, the brand of the vending machine) and identify the information you actually need.
- **Pseudocode** is a precise plan in plain language. It should be clear enough that someone else could follow it without asking you questions.
- **Validation:** trace your algorithm by hand on a small example, including an unusual one, before trusting it.
- Turtle commands: `forward(n)`, `backward(n)`, `right(deg)`, `left(deg)`, `penup()`, `pendown()`, `color(...)`. Turns are **relative** to the direction the turtle is facing now.
- To get back to the starting direction, the turns must add up to 360°. That is why a regular shape with *n* sides turns 360 ÷ *n* degrees at each corner.

```python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")

# Turtle graphics for Colab
try:
    from ColabTurtlePlus.Turtle import *
except ImportError:
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "ColabTurtlePlus"])
    from ColabTurtlePlus.Turtle import *
```
