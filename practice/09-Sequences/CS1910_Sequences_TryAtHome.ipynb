## What this workbook reviews

A **sequence** is an ordered collection, and a string is a sequence of characters. Each character has a position (an **index**), starting at 0. You can pick out one character with an index, a piece of the string with a **slice**, or visit every character with a loop. Strings are **immutable**: you never change a string, you build a new one. This workbook practises indexing, slicing, traversal, string methods and immutability.

### Key points to take away

- Indexes run from `0` to `len(s) - 1`. Negative indexes count from the end: `s[-1]` is the last character. An index past the end is an `IndexError`.
- A slice `s[start:stop:step]` **excludes** `stop`, just like `range`. Missing values take defaults: `s[:3]` is the first three characters, `s[3:]` is everything from index 3 on, and `s[::-1]` is the string reversed. Slices never raise `IndexError`.
- Two ways to traverse: `for ch in s:` when you only need the characters, and `for i in range(len(s)):` when you also need the positions (for example, to look at neighbouring characters).
- Strings are **immutable**: `s[0] = "X"` is a `TypeError`. Build a new string instead (`"X" + s[1:]`), and assign it if you want to keep it.
- String methods **return a new string** and leave the original alone: `s.upper()` does not change `s`. Write `s = s.upper()` to keep the result.
- Useful tools: `s.find(sub)` gives the index of `sub`, or `-1` if it is missing. `sub in s` is `True` or `False`. `s.strip()` removes spaces from both ends. `s.replace(old, new)` swaps every `old` for `new`.
- **Accumulator pattern for strings:** start with `result = ""` and add characters with `result = result + ch`.

```python
#@title ▶️ Run this cell first
#@markdown This cell sets up two helpers, `check` and `check_case`, that tell you
#@markdown whether your answer is right. You do not need to understand how they work.
import math

_cases_passed = {}


def _same(got, expected):
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
    return type(got) == type(expected) and got == expected


def check(got, expected, label="Answer"):
    if _same(got, expected):
        print(f"✅ {label}: {got!r}")
    elif got == expected:
        print(f"❌ {label}: the value looks right, but it is a {type(got).__name__}"
              f" and should be a {type(expected).__name__}")
    else:
        print(f"❌ {label}: got {got!r}, expected {expected!r}")


def check_case(name, case, got, expected_by_case):
    if case not in expected_by_case:
        print(f"ℹ️ {case!r} is not one of the test cases for {name}. Try one of: "
              + ", ".join(repr(c) for c in expected_by_case))
        return
    passed = _cases_passed.setdefault(name, set())
    check(got, expected_by_case[case], f"{name} with {case!r}")
    if _same(got, expected_by_case[case]):
        passed.add(case)
    still = [repr(c) for c in expected_by_case if c not in passed]
    if still:
        print(f"   {len(passed)} of {len(expected_by_case)} test cases passed. "
              f"Change the input and run again. Still to try: {', '.join(still)}")
    else:
        print(f"   🎉 All {len(expected_by_case)} test cases passed!")


print("✅ Practice checker ready.")
```
