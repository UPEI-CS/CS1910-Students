```python
# Adding Items to Lists
# Creates an empty list
my_list = [] # []
# Mutation Methods for Adding to List
#Append - Add to empty list
my_list.append(1) # [1]
# Append - Add to end of list
my_list.append(23) # [1,23]
# Append - Add to end of list
my_list.append("!") # [1, 23, "!"]
# Insert - Add specific value at specific index
my_list.insert(1, "orange") #[1, "orange", 23, "!"]
# Insert - Add to the front of the list
my_list.insert (0, "banana") # #["banana", 1, "orange", 23, "!"]
# Insert - Add specific value at end of list specific index
# Equivalent to .append(23)
my_list.insert(5, 23) #["banana", 1, "orange", 23, "!", 23]
# Insert - Add specific value well beyond the end of the list - goes to end of list
my_list.insert(99, "apple") #["banana", 1, "orange", 23, "!", 23, "apple"]
# Removing Items from Lists
# Remove the first instance of an item from the lists
my_list.remove(23) #["banana", 1, "orange", "!", 23, "apple"]
# Remove apple and store for later
last_item = my_list.pop() #["banana", 1,"orange", "!", 23]
print(last_item) # apple
# Pop - Optional parameter to remove index
# Remove first item banana for later
first_item = my_list.pop(0) #[1,"orange", "!", 23]
print(first_item) # banana
# Delete an index without knowing the value contained within
del my_list[0] #["orange", "!", 23]
# Delete a range of indices
del my_list[0:2] #Removes inclusive 0 and exclusive 2 - [23]

names_1 = ["Clark", "Diana", "Bruce"]
names_2 = names_1 # names_2 = ["Clark", "Diana", "Bruce"]
print(f"Both Lists\nList 1:\n {names_1} \nList 2:\n {names_2}")
names_1[2] = "Oliver" #Both variables point at a list that is now ["Clark, "Diana", "Oliver"]
print(f"Both Lists\nList 1:\n {names_1} \nList 2:\n {names_2}")
```

```python
# list methods
my_list = [1, 2, 75, 3, 3, 4]

#length of the list
print(f"Whole List:{my_list}")
print(f"Length: {len(my_list)}")

#sort the list in place
my_list.sort()
print(f"Sorted:{my_list}")
# Sort does not return a value
print(f"Sort method:{my_list.sort()}")

#reverse the contents of the list
my_list.reverse()
print(f"Reversed:{my_list}")

# count the occurences of
my_list.count(3)
print(f"Occurences of 3: {my_list}")

# first index position containing
my_list.index(3)
print(f"First index of 3: {my_list}")

# copy the list
cpy_list = my_list.copy()
print(f"Original:{my_list}")
print(f"Copy:{cpy_list}")

my_list[0] = 99
cpy_list[0] = 98
print(f"Updated Original:{my_list}")
print(f"Updated Copy:{cpy_list}")

# clear the contents
my_list.clear()
print(f"Cleared List - Empty list:{my_list}")
```

```python
# Ways to shallow copy
# Aliasing
original_list = [1, 2, 75, 3, 3, 4]
copied_list_1 = original_list
# Yet another way
copied_list_2 = list(original_list)
# Specify range of slice - defaults are 0 and len(list)
copied_list_3 = original_list[:]
# Alternative - imported library
import copy #imports a library with another shallow copy method
copied_list_4 = copy.copy(original_list)

print(f"Copy 1:{copied_list_1}; Copy 2:{copied_list_2}; Copy 3:{copied_list_3}; Copy 4:{copied_list_4}")
```

```python
# Deep copy method
# Creates a full copy that can be manipulated independent of the original
import copy
original_list = [1, 2, 75, 3, 3, 4]
inner_list = ["C", "D", "B"]
original_list.insert(0,inner_list) #insert inner list
deep_copied_list = copy.deepcopy(original_list) #deep copy by method
ol_inner_list = original_list[0] # get the inner list from original
dcl_inner_list = deep_copied_list[0] #get the inner list from deep copy
ol_inner_list[2] = "H" # Update the original inner list via alias
dcl_inner_list[2] = "O" # Update the copy inner list via alias
inner_list[0] = "Z" # update the original inner list
deep_copied_list[0][1] = "B" # update copy inner list via indexing
print(original_list[0]) # ['Z', 'D', 'H']
print(deep_copied_list[0]) # ['C', 'B', 'O']
print(original_list) # [['Z', 'D', 'H'], 1, 2, 75, 3, 3, 4]
print(deep_copied_list) # [['C', 'B', 'O'], 1, 2, 75, 3, 3, 4]

#Notes on 2-dimensional Matrices
# Recall Strings
name = "Stephanie"
# Offset [3] accesses string "p"
print(name[3]) # p
# Offset [0] is applied to string "p"
print(name[3][0]) #  p
# Similar for lists
# Offset [0] accesses the pointer to the list
print(deep_copied_list[0]) # ['C', 'B', 'O']
# Offset [1] is applied to the list accessed
print(deep_copied_list[0][1]) # B
```

```python
# Iterating
# Can iterate the same way
my_list = [1, 2, 75, 3, 3, 4]

# Access by value - no access to index information
for item in my_list:
    # use of second parameter to remove the default \n
    print(item, end=" ")
print() # used to print a single newline at end of loop

# Access by index - access to both value and index
for i in range(len(my_list)):
    print(my_list[i], end = " ")
print()
```

```python
# Concatenating Two lists
names_1 = ["Clark", "Diana", "Bruce"]
names_2 = ["Conner", "Donna", "Tim"]
all_names = names_1 + names_2
print(all_names) # all immutable types
#['Clark', 'Diana', 'Bruce', 'Conner', 'Donna', 'Tim']
names_2[2] = ["Richard","Tim","Damien"] # make names 2 have a nested list
all_names = names_1 + names_2 # concatenate again
print(all_names)
# ['Clark', 'Diana', 'Bruce', 'Conner', 'Donna', ['Richard', 'Tim', 'Damien']]
names_2[2][2] = "Stephanie" # update last item of inner list
print(all_names)
# ['Clark', 'Diana', 'Bruce', 'Conner', 'Donna', ['Richard', 'Tim', 'Stephanie']]
```

```python
# Unpacking Nested Lists
team_1 = ["Clark", "Diana", "Bruce"]
team_2 = ["Conner", "Donna", "Tim"]
# Create 2 nested lists inside a list
teams = []
teams.append(team_1)
teams.append(team_2)
print(teams) #[['Clark', 'Diana', 'Bruce'], ['Conner', 'Donna', 'Tim']]
team_up = [*team_1, *team_2] # Unpack the nested lists inside a new list
print(team_up) # ['Clark', 'Diana', 'Bruce', 'Conner', 'Donna', 'Tim']
# Unpack the nested lists and insert some new items
big_team_up = ["Oliver,",*team_1, "Dinah", *team_2]
print(big_team_up) # ['Oliver,', 'Clark', 'Diana', 'Bruce', 'Dinah', 'Conner', 'Donna', 'Tim']
```

```python
# Creating a deck of cards
# highest to lowest rankings
suit = ["♠", "♥", "♦", "♣"]
rank = ["A", "K", "Q", "J", "10", "9", "8", "7", "6", "5", "4", "3", "2"]
#10 of hearts
ten_hearts = rank[4] + suit[1]
print(f"10 of hearts: {ten_hearts}")
# create a deck of cards with all cards grouped by suit and ordered by rank
deck = []
# using value based for loop as I do not need the index
for s in suit:
    for card in rank:
        # add it to the end of the deck
        deck.append(card + s)
print(deck) # full deck
```

```python
# Swapping Places
# Get index of first (and only) 2 of hearts
idx_2h = deck.index("2♥")
# Get index of first (and only) 2 of clubs
idx_2c = deck.index("2♣")
# Key pattern - Holding temporary variable
# Swap method 2
temp_card = deck[idx_2h] # move card 1 to temp variable
deck[idx_2h] = deck[idx_2c] # move card 2 to card 1
deck[idx_2c] = temp_card # move temp to card 2
print(deck)
# Swap back
temp_card = deck[idx_2c] # move card 1 to temp variable
deck[idx_2c] = deck[idx_2h] # move card 2 to card 1
deck[idx_2h] = temp_card # move temp to card 2
print(deck)
#Python Specific Swap Expression
deck[idx_2h], deck[idx_2c] = deck[idx_2c], deck[idx_2h]
print(deck)
```

```python
# Mutable Objects in Functions
def add_team_member(team, new_member):
    """ Adds a team member if they are not on the team.
    :param team: the list of team members
    :param add: the person to add to the team
    :return: true if added, else false
    """
    if new_member not in team:
        team.append(new_member)
        return True
    else:
        return False

original_team = ["Diana", "Arthur", "John", "Harold", "Barry"]
print(original_team) # ['Diana', 'Arthur', 'John', 'Harold', 'Barry']

add_team_member(original_team, "Clark")
add_team_member(original_team, "Bruce")
print(original_team) # ['Diana', 'Arthur', 'John', 'Harold', 'Barry', 'Clark', 'Bruce']
```
