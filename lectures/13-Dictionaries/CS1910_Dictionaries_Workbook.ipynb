```python
# Using Dictionaries
contact = {"name"    : "Clark Kent",
            "age"   : 33,
            "address" : "Metropolis" }
print(contact) # {'name': 'Clark Kent', 'age': 33, 'address': 'Metropolis'}
# Get value associated with key "age"
print(contact["age"]) # 33
print(contacts["secret_identity"]) # Produces KeyError
# Update key
contact["name"] = "Superman"
print(contact) #{'name': 'Superman', 'age': 33, 'address': 'Metropolis'}
# Add a new Key
contact["secret_identity"] = "Clark Kent"
print(contact["secret_identity"]) # Clark Kent
```

```python
# Duplicate Keys - Results in Overwrite
contacts = {"name"    : "Clark Kent",
            "age"   : 33,
            "address" : "Metropolis",
            "name"    : "Bruce Wayne",
            "age"   : "Unknown",
            "address" : "Gotham" }

# Reusing keys in creation results in overwriting values
print(contacts) #{'name': 'Bruce Wayne', 'age': 'Unknown', 'address': 'Gotham'}
```

```python
#List of Dictionaries
# Can hold multiple dictionaries
contacts = [ { "name" : "Clark Kent",
               "age"  : 33,
               "address" : "Metropolis" },
             { "name" : "Bruce Wayne",
               "age"  : "Unknown",
               "address" : "Gotham" }
           ]
print(len(contacts)) # Length of list contacts
print(len(contacts[0])) # Number of items in dictionary
print(contacts[0]) # dictionary at 0
print(contacts[1]) # dictionary at 1

# Nested Dictionaries
contact =   {"Batman":    {
                            "name" : "Bruce Wayne",
                            "age"  : "Unknown",
                            "address" : "Gotham",
                            "sidekick": {"Robin" :   {
                                                        "name" : "Richard Grayson",
                                                        "age"  : 16,
                                                        "address" :"Bludhaven"
                                                    }
                                        }
                        }
            }

# Resolve each key in precedence from left to right like lists
print(contact["Batman"]["sidekick"]) # {'Robin': {'name': 'Richard Grayson', 'age': 16, 'address': 'Bludhaven'}}
print(contact["Batman"]["sidekick"]["Robin"]["name"]) #Richard Grayson
```

```python
# .pop method on dictionary
hero_1 = {"name"    : "Clark Kent",
            "age"   : 33,
            "address" : "Metropolis",
            "email" : "ckent@dailyplanet.com"}
age = hero_1.pop("age") #get the value for age and remove key:value pair
print(age)  # 33
del hero_1["address"]
print(hero_1) # {'name': 'Clark Kent', 'email': 'ckent@dailyplanet.com'}
#pop "email" but supply a default
#the default is returned if email is not a valid key in the dictionary
hero_2 = {  "name" : "Bruce Wayne",
            "age"  : "Unknown",
            "address" : "Gotham" }
email = hero_2.pop("email", "Unknown")
print(email) #Unknown
print(hero_2) #{'name': 'Bruce Wayne', 'age': 'Unknown', 'address': 'Gotham'}
```

```python
# Keys Method - Creates a 'view" of the keys
my_dict = {'a': 1, 'b': 2, 'c': 3}
# Create a view object for keys
keys_view = my_dict.keys()
# keys_view can "see" all of the keys
print(keys_view) #dict_keys(['a', 'b', 'c'])
print(f"Initial keys view: {keys_view}")
# Initial keys view: dict_keys(['a', 'b', 'c'])

# looks like a list when printed - but is not a list
# print(keys_view[1]) #fails with TypeError
# keys_view cannot be modified directly
# keys_view.append('d') # fails - not a list

# Modify the dictionary
my_dict['d'] = 4
# After modifying the dictionary, the keys_view variable
print(f"Keys view after dictionary modification: {keys_view}")
# Keys view after dictionary modification: dict_keys(['a', 'b', 'c', 'd'])

# Same thing with .values()
values_view = my_dict.values()
print(values_view)
my_dict['e'] = 5
print(f"Values view after dictionary modification: {values_view}")

# .items() returns an 'iterable'
# Iterable - means it is a data structure that can be iterated over
items_view = my_dict.items()
print(f"Items view (tuples)): {items_view}")

```

```python
# Iterating over dictionaries
my_dict = {'a': 1, 'c': 3, 'b': 2}
my_dict['d'] = 4
#Iterates over the keys
#(in the order they were inserted)
for k in my_dict:
    print(k, end=" ")
print() # a c b d
#Iterates over the keys
#Retrieve keys view
for k in my_dict.keys():
    print(k, end=" ")
print() # a c b d
#Iterates over the values
for v in my_dict.values():
    print(v, end=" ")
print() # 1 3 2 4
# iterable - data structure that can be iterated over
# .items() - returns an iterable of tuples
# Iterates over tuples
for k,v in my_dict.items():
    print(k,v, end= " ")
print() # a 1 c 3 b 2 d 4
for item in my_dict.items():
    print(item[0], item[1], end= " ")
print() # a 1 c 3 b 2 d 4
#Iterate over the keys but access the values
for k in my_dict:
    print(k, my_dict[k], end= " ")
print() # a 1 c 3 b 2 d 4
```

```python
# Class participation
# Which approach adds 10 to each value in a dictionary

my_dict={ 'a':1, 'b':2, 'c':3}

#A
for v in my_dict.values():
    v+=10
#B
for k in my_dict:
    my_dict[k] += 10

#C
for k,v in my_dict.items():
    my_dict[k] = v + 10
#D
my_dict = {k:v+10 for k,v in my_dict.items()}

print(my_dict)
```

You are given a list that represents a cipher.
* Even positions contain symbols
* Odd positions contain meanings
* Assume one UNICODE symbol for each ASCII capital letter character

**Task:** Your task is to translate this list into a dictionary so it can be used to decode messages.
**Task:** Create a function called transform_cipher that takes in a list and returns a dictionary


```python

# Original Cipher list
cipher_list = [
    "@", "A",
    "#", "B",
    "$", "C",
    "%", "D",
    "&", "E",
    "*", "F",
    "+", "G",
    "=", "H",
    "?", "I",
    "!", "J",
    "^", "K",
    "~", "L",
    "<", "M",
    ">", "N",
    "/", "O",
    "|", "P",
    "{", "Q",
    "_", "R",
    "-", "S",
    ":", "T",
    ";", "U",
    ",", "V",
    ".", "W",
    "`", "X",
    "'", "Y",
    "\"", "Z"
]

# Translate into a dictionary
cipher_dict = {}

for i in range(0, len(cipher_list), 2):
    symbol = cipher_list[i]
    meaning = cipher_list[i + 1]
    cipher_dict[symbol] = meaning
print(cipher_dict)

print(cipher_dict["~"])   # L
print(cipher_dict["/"])   # O
print(cipher_dict["~"])   # L

```

```python
# Cipher transformation as a function

def transform_cipher(cipher_list):
    """
    Takes a list-based cipher and returns a dictionary-based cipher.
    Even indices are symbols, odd indices are meanings.
    """
    cipher_dict = {}

    for i in range(0, len(cipher_list), 2):
        symbol = cipher_list[i]
        meaning = cipher_list[i + 1]
        cipher_dict[symbol] = meaning

    return cipher_dict


def encode_msg(symbol_msg):
    """
    Takes a string of symbols and encodes it into letters
    using a list-based cipher.
    """
    cipher_list = [
        "@", "A", "#", "B", "$", "C", "%", "D", "&", "E",
        "*", "F", "+", "G", "=", "H", "?", "I", "!", "J",
        "^", "K", "~", "L", "<", "M", ">", "N", "/", "O",
        "|", "P", "{", "Q", "_", "R", "-", "S", ":", "T",
        ";", "U", ",", "V", ".", "W", "`", "X", "'", "Y",
        "\"", "Z"
    ]

    result = ""
    for char in symbol_msg:
        for i in range(1, len(cipher_list), 2):  # step through the letter positions
            if cipher_list[i] == char:
                result += cipher_list[i - 1]      # grab the symbol just before it
    return result



def decode_msg(letter_msg):
    """
    Takes a string of letters and decodes it back into symbols
    using a dictionary-based cipher.
    """

    cipher_dict = {
        "@": "A", "#": "B", "$": "C", "%": "D", "&": "E",
        "*": "F", "+": "G", "=": "H", "?": "I", "!": "J",
        "^": "K", "~": "L", "<": "M", ">": "N", "/": "O",
        "|": "P", "{": "Q", "_": "R", "-": "S", ":": "T",
        ";": "U", ",": "V", ".": "W", "`": "X", "'": "Y",
        "\"": "Z"
    }

    result = ""
    for char in letter_msg:
        result += cipher_dict[char]
    return result




encoded_str = encode_msg("ALO")
print(encoded_str)
# @~/
decoded_str = decode_msg("@~/")
print(decoded_str)
# ALO 
```

# Find a Partner

You have a 2-dimensional image where there is a boundary of some kind.
Example:
An outline of your favourite character
A map with a coastline boundary
You have 3 characters that can represent inside the boundary, outside the boundary, and the boundary itself:
* @ - Boundary
* \# - outside the boundary
* . – Inside the boundary

Now assume that you have encoding the picture so that @ is represented by 1, inside the boundary is 2 and outside the boundary is 1
* How would you store the picture?
* How would you store the translation?
* How would you store the mapping from one to another?
* Sketch the algorithm for the original creation of the encoding
* Translate that encoding into Python.


```python
# Find a partner solution

pixel_map = {
    "@": 1,
    ".": 0,
    "#": 2
}


image = [
    ["@", "@", "#", "#", "#"],
    ["@", ".", ".", "#", "#"],
    ["@", ".", ".", ".", "#"],
    ["@", "@", "@", ".", "."]
]

print(image)

numeric_image = []

for row in image:
    new_row = []
    for symbol in row:
        new_row.append(pixel_map[symbol])
    numeric_image.append(new_row)

print(numeric_image)
```

```python
## Dictionary Comprehensions

#new_dict = {key_expression : value_expression for item in iterable}

numbers = [1, 2, 3, 4]
squared_dict = {num: num ** 2 for num in numbers}
# Result: {1: 1, 2: 4, 3: 9, 4: 16}
print(squared_dict)

#optional condition
original_dict = {'a': 10, 'b': 25, 'c': 5, 'd': 30}
filtered_and_doubled = {k: v * 2 for k, v in original_dict.items() if v > 15}
# Result: {'b': 50, 'd': 60}

print(filtered_and_doubled)
```
