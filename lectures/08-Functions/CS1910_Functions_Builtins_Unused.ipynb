# 2. A Range of numbers
When describing a range of numbers, different types of brackets are used to indicate whether the endpoints are included or excluded:
Inclusive Range:  
*   Inclusive Range: [x..y] — This means the range includes both endpoints x and y. In other words, the values are between x and y, including both x and y.
*   Left-Exclusive, Right-Inclusive Range: (x..y] — This means the range includes all values between x and y, excluding x but including y.
*   Right-Inclusive, Right-Exclusive Range: [x..y) — This means the range includes all values between x and y, including x but excluding y.
*   Exclusive Range: (x..y) — This means the range exclude both endpoints x and y. In other words, the values are between x and y, excluding both x and y.

````python
#@title 3.1 Verify Your Program Creation
#@markdown Answer the following questions based on the code you wrote in `move.py` and `stop.py` files

#@markdown When you run the cell, you will see some Python statements followed by a text box. Enter your answers in the text box provided.

import ipywidgets as widgets
import IPython.display as ipd

# Question data
questions = [
    {
        "number": 1,
        "type": "Submit",
        "question": "Record the output obtained from running the `move.py` file.",
        "sub_questions": ["Line 1:", "Line 2:", "Line 3"]
    },
    {
        "number": 2,
        "type": "Submit",
        "question": "Consider each line of the output from the run of `stop.py`. Which line in `move.py` does each correspond to? Or which in `stop.py` is it generated by?",
        "sub_questions": ["Line 1:", "Line 2:", "Line 3", "Line 4"]
    }
]

# Create answer widgets for questions
answer_widgets = []
question_containers = []

for i, q in enumerate(questions):
    # Create question header using markdown
    question_markdown = f"""#### Question {q['number']}:
{q['question']}"""

    question_display = widgets.Output()
    with question_display:
        ipd.display(ipd.Markdown(question_markdown))


    # Create multiple small text widgets for sub-questions
    sub_widgets = []
    for sub_q in q["sub_questions"]:
        label = widgets.HTML(f"<b>{sub_q}</b>")
        text_input = widgets.Text(
            placeholder='Enter your answer...',
            layout=widgets.Layout(width='30%', margin='0 0 5px 0')
        )
        sub_widgets.append(widgets.HBox([label, text_input]))
    widget = widgets.VBox(sub_widgets)

    answer_widgets.append(widget)

    # Create container for this question
    question_container = widgets.VBox([
        question_display,
        widget
    ], layout=widgets.Layout(
        width='95%',
        max_width='800px',
        margin='0 0 25px 0'
    ))

    question_containers.append(question_container)

# Submit and Edit buttons
submit_button = widgets.Button(
    description='Submit All Answers',
    button_style='success',
    layout=widgets.Layout(width='200px', margin='10px 5px 0 0')
)

edit_button = widgets.Button(
    description='Edit Answers',
    button_style='warning',
    layout=widgets.Layout(width='150px', margin='10px 5px 0 0')
)

# Reveal answers button
reveal_button = widgets.Button(
    description='Reveal Answers',
    button_style='info',
    layout=widgets.Layout(width='150px', margin='10px 0 0 0')
)

# Output area for displaying submitted answers
output_area = widgets.Output()

def submit_answers(button):
    with output_area:
        ipd.clear_output(wait=True)
        print("📝 Submitted Questions:")
        print("=" * 80)

        for i, (q, widget_vbox) in enumerate(zip(questions, answer_widgets)):
            # widget_vbox contains multiple HBox sub-widgets
            answers = []
            for hbox in widget_vbox.children:
                # Each HBox: [label, Text]
                text_input = hbox.children[1]
                answers.append(text_input.value.strip() if text_input.value.strip() else "(no answer provided)")

            print(f"\n Question {q['number']}:")
            print(f"Q: {q['question']}")
            for j, ans in enumerate(answers):
                print(f"{q['sub_questions'][j]} {ans}")
            print("-" * 60)

        print("\n✅ All answers saved! You can edit them anytime using the 'Edit Answers' button.")

    # Hide form and show edit/reveal buttons
    form_container.layout.display = 'none'
    submit_button.layout.display = 'none'
    edit_button.layout.display = 'block'
    reveal_button.layout.display = 'block'

def edit_answers(button):
    # Show form and hide edit/reveal buttons
    form_container.layout.display = 'block'
    submit_button.layout.display = 'block'
    edit_button.layout.display = 'none'
    reveal_button.layout.display = 'none'

    with output_area:
        ipd.clear_output(wait=True)
        print("📝 Edit mode activated - make your changes and click 'Submit All Answers' again.")

def reveal_answers(button):
    # Replace this answer_key with the correct answers for your questions
    answer_key = [
        ["in move: __name__ == __main__", "will always execute: angle == 68", "only if True: angle == -39"],  # Q1
        ["Line 7", "Line 8", "Line 3", "Line 4"]  # Q2
    ]

    with output_area:
        ipd.clear_output(wait=True)
        print("📚 ANSWER KEY:")
        print("=" * 80)

        for i, (q, widget_vbox) in enumerate(zip(questions, answer_widgets)):
            # Get student answers
            student_answers = []
            for hbox in widget_vbox.children:
                text_input = hbox.children[1]
                student_answers.append(text_input.value.strip() if text_input.value.strip() else "(no answer provided)")

            correct_answers = answer_key[i]

            print(f"\n Question {q['number']}:")
            print(f"Q: {q['question']}")
            for j, (student, correct) in enumerate(zip(student_answers, correct_answers)):
                print(f"{q['sub_questions'][j]} Your Answer: {student} | Model Answer: {correct}")
            print("-" * 80)

        print("\n📝 Note: These are model answers. Your responses may vary in wording while still demonstrating understanding.")

# Connect button functions
submit_button.on_click(submit_answers)
edit_button.on_click(edit_answers)
reveal_button.on_click(reveal_answers)

# Initially hide edit and reveal buttons
edit_button.layout.display = 'none'
reveal_button.layout.display = 'none'

# Create main containers
form_container = widgets.VBox(
    question_containers,
    layout=widgets.Layout(
        width='95%',
        max_width='800px',
        margin='0 0 0 0'
    )
)

button_container = widgets.HBox(
    [submit_button, edit_button, reveal_button],
    layout=widgets.Layout(
        width='95%',
        max_width='800px',
        margin='10px 0 0 0'
    )
)

# Display everything at once to prevent scrolling
all_content = widgets.VBox([
    form_container,
    button_container,
    output_area
], layout=widgets.Layout(
    width='95%',
    max_width='800px',
    margin='0 0 0 0'
))

ipd.display(all_content)
````
