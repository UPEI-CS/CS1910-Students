````python
# Find a Partner - Quadrant Image Problem
import matplotlib.pyplot as plt

pixel = 0
# 3 x 3 image
image = [ [pixel, pixel, pixel],
          [pixel, pixel, pixel],
          [pixel, pixel, pixel]
        ]

# create a 128 x 64 black image
#an empty list in which to store 128 other lists, each with 64 0 values
image = []

width = 64
height = 128

# Pattern we need for display:
# Height: 128; Width: 64
# Two options - a list storing:
#   1) 128 inner lists with 64 values (128 rows, 64 columns)
#   2) 64 inner lists with 128 values (64 rows, 128 columns)
# It is actually the first which is dictated by the display module pyplot
# Example of generalization hiding important information where we have to check

# Solution 1: Nested Loop
# Fill the screen with black

for i in range(height):
    row = []
    for j in range(width):
        row.append(0)
    image.append(row)

#resulting image is 128 lists with 64 0's
# Horizontal white line
#List 0:    0,  0,  0,  0
#List 1:    0,  0,  0,  0
#List 2:  255,255,255,255
#List 3:    0,  0,  0,  0

# Go to the middle list, and fill it with 255
for i in range(width):
    image[64][i] = 255

# now we will find that one of the columns is all 255
#print(image)

# Then we need to do the vertical
# We need to update every column so that the one in the middle is white
#   for every column
#       update index 64/2 to white
for j in range(height):
    image[j][32] = 255

# now we have 1 column of all 255, and all others have 1 255 in them
#print(image)

# this code can help display an image
#once you have some values inside image uncomment the below 3 lines
plt.imshow(image,cmap='gray')
plt.axis('off') # Hide axes for cleaner display
plt.show()

# Shorter code with expressions
image = []

for i in range(height):
    if i == height//2: # i is an integer for comparison we must create an integer
        image.append([255] * width)
    else:
        image.append([0] * (width//2) + [255] + [0] * (width - width//2 - 1))

plt.imshow(image,cmap='gray')
plt.axis('off') # Hide axes for cleaner display
plt.show()

# Comprehension Example
image = [
    [255 if i == height // 2 or j == width // 2 else 0
     for j in range(width)]
    for i in range(height)
]

plt.imshow(image,cmap='gray')
plt.axis('off') # Hide axes for cleaner display
plt.show()

````

````python
# Find a Partner — Gradient Image
import matplotlib.pyplot as plt

# We are still creating a 128 x 64 grayscale image
# Height corresponds to rows, width corresponds to columns
width = 64
height = 128

image = []

# We build the image one row at a time
for i in range(height):
    row = []

    # For each row, we calculate pixel values across the width
    for j in range(width):

        # The gradient should run left to right
        # So the column index (j) controls the pixel value

        # j ranges from 0 to width-1
        # We need to map that range onto pixel values 0 to 255
        pixel_value = int((j / (width - 1)) * 255)

        row.append(pixel_value)

    image.append(row)

# Display the resulting image
plt.imshow(image, cmap='gray')
plt.axis('off')
plt.show()
````
